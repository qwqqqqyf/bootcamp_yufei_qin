## 2. Compare Loop vs Vectorization

```python
# TODO: Compare speed using %timeit
arr = np.arange(10000)

def loop_operation(arr):
    result = []
    for x in arr:
        result.append(x * 2)
    return result

def vectorized_operation(arr):
    return arr * 2

print("Loop:")
%timeit loop_operation(arr)

print("Vectorized:")
%timeit vectorized_operation(arr)
```
